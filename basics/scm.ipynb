```python
import numpy as np
import stats
```

```python
class bookSCM:
    def __init__(self,seed=None):
        self.seed=seed
        self.u0=stats.uniform()
        self.u1=stats.norm()

    def sample(self,n=1):
        if self.seed is not None:
            np.random.seed(self.seed)
        u0=self.u0.rvs(n)
        u1=self.u1.rvs(n)
        a=u0>0.61
        b=(a+0.5*u1)>0.5
        return a,b

scm=bookSCM(seed=42)
buy_book_a,buy_book_b=scm.sample(n=100)
```

```python
class CFSCM:
    def abduct(self,t,y):
        return (t+y-1)/(2*t-1)
    def modify(self,t):
        return lambda u: t*u+ (t-1)*(1-u)
    def predict(self,t,u):
        return self.modify(t)(u)

coffee=CFSCM()
t=1
y=1

u=coffee.abduct(t,y)
print(u)

print(coffee.predict(t=0,u=1))
```
